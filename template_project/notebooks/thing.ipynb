```python
import geopandas as gpd
from gerrychain import Graph, Partition
```

```python
gdf = gpd.read_file("../data/pa_districtr_vtd_view_v2.gpkg", layer="pa_districtr_vtd_view_v2")
gdf
```

```python
graph = Graph.from_geodataframe(gdf.to_crs("EPSG:5070"))
```

```python
partition = Partition.from_random_assignment(graph, n_parts=50, epsilon=0.01, pop_col="total_pop_20")
```

```python
gdf["seed_plan"] = partition.assignment_vector
```

```python
graph_2 = Graph.from_geodataframe(gdf.to_crs("EPSG:5070"))
```

```python
graph_2.to_json("../JSON_dualgraphs/pa_dualgraph.json")
```
